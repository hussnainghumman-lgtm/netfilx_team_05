# Movie Database Analysis

## 1. Objectives

- Explore the movie database.
- Analyse movie ratings.
- Identify the most popular movie genres.
- Understand user viewing behaviour.
- Identify the most active users.
- Visualise the results using charts.


# 2. Import Libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

# 3. Load Data

```python

users = pd.read_csv("users.csv")
movies = pd.read_csv("movies.csv")
watch_history = pd.read_csv("watch_history.csv")

print("All CSV files loaded successfully!")

print("Total users:", len(users))
print("Total movies:", len(movies))
print("Total watch records:", len(watch_history))
```

# 4. Explore the Data

```python


print("USERS TABLE")
display(users.head())

print("MOVIES TABLE")
display(movies.head())

print("WATCH HISTORY TABLE")
display(watch_history.head())
```

# 5. Data Cleaning

```python

# Check missing values
print("Missing values in Users:")
print(users.isnull().sum())

print("\nMissing values in Movies:")
print(movies.isnull().sum())

print("\nMissing values in Watch History:")
print(watch_history.isnull().sum())

# Remove duplicate records
users = users.drop_duplicates()
movies = movies.drop_duplicates()
watch_history = watch_history.drop_duplicates()

# Convert date column into date format
watch_history["date"] = pd.to_datetime(watch_history["date"])

print("\nData cleaning completed successfully!")
```

# 6. Merge Tables

```python

# Merge Watch History with Movies
analysis_df = watch_history.merge(
    movies,
    on="movie_id",
    how="left"
)

# Merge with Users
analysis_df = analysis_df.merge(
    users,
    on="user_id",
    how="left"
)

print("Tables merged successfully!")

display(analysis_df.head(10))
```

# 7. Descriptive Statistics

```python

print("Average movie rating:")
print(analysis_df["rating"].mean())

print("\nHighest rating:")
print(analysis_df["rating"].max())

print("\nLowest rating:")
print(analysis_df["rating"].min())

print("\nRating summary:")
display(analysis_df["rating"].describe())
```

# 8. Movie Rating Analysis

```python

movie_ratings = (
    analysis_df.groupby("title")["rating"]
    .mean()
    .sort_values(ascending=False)
)

print("Average rating for each movie:")
display(movie_ratings)
```

# 9. Genre Analysis

```python

genre_counts = analysis_df["genre"].value_counts()

print("Number of watches by genre:")
display(genre_counts)

# Create a bar chart
plt.figure(figsize=(9, 5))

genre_counts.plot(kind="bar")

plt.title("Popular Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Watches")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# 10. User Activity Analysis

```python

user_activity = (
    analysis_df.groupby("name")["movie_id"]
    .count()
    .sort_values(ascending=False)
)

print("Movies watched by each user:")
display(user_activity)

# Create chart
plt.figure(figsize=(10, 5))

user_activity.plot(kind="bar")

plt.title("User Viewing Activity")
plt.xlabel("User Name")
plt.ylabel("Number of Movies Watched")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# 11. Movie Release Year Analysis

```python

year_counts = movies["year"].value_counts().sort_index()

print("Movies by release year:")
display(year_counts)

plt.figure(figsize=(10, 5))

year_counts.plot(kind="bar")

plt.title("Movies by Release Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

# 12. Rating Distribution

```python

rating_counts = analysis_df["rating"].value_counts().sort_index()

print("Rating distribution:")
display(rating_counts)

plt.figure(figsize=(8, 5))

rating_counts.plot(kind="bar")

plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Ratings")
plt.tight_layout()
plt.show()
```

# 13. Final Project Summary

```python

print("========== MOVIE DATABASE SUMMARY ==========")

print("Total Users:", users["user_id"].nunique())

print("Total Movies:", movies["movie_id"].nunique())

print("Total Watches:", len(analysis_df))

print("Average Rating:", round(analysis_df["rating"].mean(), 2))

print("Most Popular Genre:", genre_counts.idxmax())

print("Most Active User:", user_activity.idxmax())

print("Highest Rated Movie:", movie_ratings.idxmax())

print("============================================")
```

# 14. Save Final Dataset

```python

analysis_df.to_csv("movie_analysis_combined.csv", index=False)

print("Final analysis dataset exported successfully!")
```
